```python
import control as ctrl
import numpy as np
import matplotlib.pyplot as plt
```

```python
# --- System Definition ---
# Transfer Function: G(s) = 1 / (s^4 + 2s^3 + 3s^2 + 2s)
# Note: The '0' at the end of 'den' signifies a pole at the origin (s=0).
num = [1]
den = [1, 2, 3, 2, 0]

sys = ctrl.TransferFunction(num, den)

print("System Transfer Function:")
print(f"G(s) = {sys}")
```

```python
# --- Stability & Characteristics ---
poles = ctrl.poles(sys)
zeros = ctrl.zeros(sys)

print(f"\nPoles: {poles}")
print(f"Zeros: {zeros}")
```

```python
# Plot 1: Step Response
# Because of the pole at s=0, the system acts as an integrator (ramp output).
plt.figure(figsize=(10, 5))
t, y = ctrl.step_response(sys)
plt.plot(t, y, color='#0D9488', linewidth=2)
plt.title('System Step Response (Integrator Behavior)', fontsize=14, fontweight='bold')
plt.xlabel('Time (s)')
plt.ylabel('Output')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Plot 2: Pole-Zero Map
plt.figure(figsize=(8, 6))
ctrl.pzmap(sys, grid=True)
plt.title('Pole-Zero Map of the System', fontsize=14, fontweight='bold')
plt.show()
```
